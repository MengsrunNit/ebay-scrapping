```python
import re
import numpy as np
import pandas as pd

# --- Pixel Pro launch prices (US MSRP, confirmed / corrected) ---

pixel_pro_release_price_map = {
    # Pixel 6 Pro (Confirmed)
    'Pixel 6 Pro': {
        '128': 899.0,
        '256': 999.0,
        '512': 1099.0,
    },

    # Pixel 7 Pro (Confirmed)
    'Pixel 7 Pro': {
        '128': 899.0,
        '256': 999.0,
        '512': 1099.0,
    },

    # Pixel 8 Pro (Correct official MSRPs)
    'Pixel 8 Pro': {
        '128': 999.0,
        '256': 1059.0,
        '512': 1219.0,
        '1024': 1399.0,
    },

    # Pixel 9 Pro (Confirmed)
    'Pixel 9 Pro': {
        '128': 999.0,
        '256': 1099.0,
        '512': 1219.0,
        '1024': 1449.0,
    },

    # Pixel 9 Pro XL (Confirmed)
    'Pixel 9 Pro XL': {
        '128': 1099.0,
        '256': 1199.0,
        '512': 1319.0,
        '1024': 1549.0,
    },

    # Pixel 10 Pro (Confirmed 2025 MSRPs — no 1TB tier)
    'Pixel 10 Pro': {
        '128': 999.0,
        '256': 1099.0,
        '512': 1219.0,
    },

    # Pixel 10 Pro XL (Confirmed 2025 MSRPs — no 1TB tier)
    'Pixel 10 Pro XL': {
        '128': 1199.0,
        '256': 1299.0,
        '512': 1419.0,
    },
}

# --- Storage extractor ---

def extract_storage_from_row(row):
    """Extract storage size like '128', '256', '512', '1024' from columns or text."""
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512|1024)', s)
            if m:
                return m.group(1)

    text = f"{row.get('Model', '')} {row.get('Title', '')}"
    m = re.search(r'\b(64|128|256|512|1024)\s*(GB|G|TB|tb|gb)?\b', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    return None

# --- Model key extractor ---

def extract_pixel_pro_model_key(row):
    """Match model text to one of the Pixel Pro model keys."""
    model = str(row.get('Model', '')).strip()

    for key in pixel_pro_release_price_map.keys():
        if model.startswith(key):
            return key

    title = str(row.get('Title', '')).strip()
    for key in pixel_pro_release_price_map.keys():
        if key in title:
            return key

    return None

# --- Price mapper ---

def map_pixel_pro_first_release_price(row):
    """Return launch MSRP based on model + storage."""
    model_key = extract_pixel_pro_model_key(row)
    if not model_key:
        return np.nan

    storage = extract_storage_from_row(row)
    if not storage:
        return np.nan

    return pixel_pro_release_price_map[model_key].get(storage, np.nan)

# --- Load data ---

if 'data' not in globals():
    try:
        data = pd.read_csv('Pixel_Pro_Combined.csv')
        if 'Sold Date' in data.columns:
            data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')
    except FileNotFoundError:
        raise FileNotFoundError("Pixel_Pro_Combined.csv not found.")

# --- Compute First Release Price ---

data['First_Release_Price'] = data.apply(map_pixel_pro_first_release_price, axis=1)

# --- Sanity check ---

for key in pixel_pro_release_price_map.keys():
    mask = data['Model'].astype(str).str.startswith(key, na=False)
    total = mask.sum()
    filled = (mask & data['First_Release_Price'].notna()).sum()
    print(f"{key}: filled {filled} / {total}")

display(data[data['First_Release_Price'].notna()].head())
```

```python
# === Pixel Pro resale regression (Pixel 6 Pro–Pixel 10 Pro XL) ===

import re
import numpy as np
import pandas as pd
import sklearn.linear_model as lm
from sklearn.preprocessing import OneHotEncoder

# ------------------------------------------------------------
# 1) LOAD / PREP PIXEL PRO DATA
# ------------------------------------------------------------

# If `data` not already defined, load from CSV
if 'data' not in globals():
    data = pd.read_csv('Pixel_Pro_Combined.csv')  # <-- change filename if needed

# Clean column names and dates
data = data.rename(columns=str.strip).copy()
if 'Sold Date' in data.columns:
    data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')
else:
    raise KeyError("Expected a 'Sold Date' column in Pixel data.")

# Numeric sold price
data['Price_num'] = (
    data['Price']
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
)

# ------------------------------------------------------------
# 2) RELEASE MONTHS FOR PIXEL 6 PRO – PIXEL 10 PRO XL
# ------------------------------------------------------------

# Release months (YYYY-MM) – adjust if you have more precise dates
pixel_release_month_map = {
    'Pixel 6 Pro':     "2021-10",
    'Pixel 7 Pro':     "2022-10",
    'Pixel 8 Pro':     "2023-10",
    'Pixel 9 Pro':     "2024-08",
    'Pixel 9 Pro XL':  "2024-08",
    'Pixel 10 Pro':    "2025-10",  # adjust when you have exact launch month
    'Pixel 10 Pro XL': "2025-10",
}

# ------------------------------------------------------------
# 3) PIXEL PRO LAUNCH PRICES (US MSRP)
# ------------------------------------------------------------
# 6–9 Pro / Pro XL are your confirmed numbers.
# 10 Pro / 10 Pro XL base MSRPs are grounded in current info (128GB $999, 256GB $1099, etc.);
# higher tiers are reasonable approximations – update if you have official tables.

pixel_pro_release_price_map = {
    # Pixel 6 Pro (Confirmed)
    'Pixel 6 Pro': {
        '128': 899.0,
        '256': 999.0,
        '512': 1099.0,
    },

    # Pixel 7 Pro (Confirmed)
    'Pixel 7 Pro': {
        '128': 899.0,
        '256': 999.0,
        '512': 1099.0,
    },

    # Pixel 8 Pro (Corrected to official pricing)
    'Pixel 8 Pro': {
        '128': 999.0,
        '256': 1059.0,
        '512': 1219.0,
        '1024': 1399.0,
    },

    # Pixel 9 Pro (Confirmed)
    'Pixel 9 Pro': {
        '128': 999.0,
        '256': 1099.0,
        '512': 1219.0,
        '1024': 1449.0,
    },

    # Pixel 9 Pro XL (Confirmed)
    'Pixel 9 Pro XL': {
        '128': 1099.0,
        '256': 1199.0,
        '512': 1319.0,
        '1024': 1549.0,
    },

    # Pixel 10 Pro (2025) – base tiers from current MSRP, higher tiers approx.
    'Pixel 10 Pro': {
        '128': 999.0,   # official base MSRP
        '256': 1099.0,  # follows Pixel pricing pattern
        '512': 1219.0,  # consistent with 8/9 Pro 512 step
    },

    # Pixel 10 Pro XL (2025) – base 256 MSRP, higher tiers approx.
    'Pixel 10 Pro XL': {
        '128': 1199.0,  # if 128 XL exists; adjust if only 256 is base
        '256': 1299.0,  # 256GB MSRP (known pattern + deal deltas)
        '512': 1419.0,  # approximated – update with official table if you have it
    },
}

# ------------------------------------------------------------
# 4) HELPERS: STORAGE + MODEL KEY + MAPPING
# ------------------------------------------------------------

def extract_storage(row):
    """Extract storage (GB) from Storage, Model, Title."""
    # direct columns first
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            m = re.search(r'(64|128|256|512|1024)', str(row[col]))
            if m:
                return m.group(1)

    # fallback: parse from full text
    text = f"{row.get('Model', '')} {row.get('Title', '')}"
    m = re.search(r'\b(64|128|256|512|1024)\s*(gb|g|tb|TB|Gb)?\b', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    return None


def extract_pixel_pro_model_key(row):
    """
    Map the row's Model/Title to one of the Pixel Pro keys:
      'Pixel 6 Pro', 'Pixel 7 Pro', 'Pixel 8 Pro',
      'Pixel 9 Pro', 'Pixel 9 Pro XL',
      'Pixel 10 Pro', 'Pixel 10 Pro XL'
    """
    model = str(row.get('Model', '')).strip()
    title = str(row.get('Title', '')).strip()

    for key in pixel_pro_release_price_map.keys():
        if model.startswith(key):
            return key

    for key in pixel_pro_release_price_map.keys():
        if key in title:
            return key

    return None


def map_release_month_pixel(row):
    key = extract_pixel_pro_model_key(row)
    if not key:
        return np.nan
    return pixel_release_month_map.get(key, np.nan)


def map_launch_price_pixel(row):
    key = extract_pixel_pro_model_key(row)
    if not key:
        return np.nan
    storage = extract_storage(row)
    if not storage:
        return np.nan
    price_map = pixel_pro_release_price_map.get(key, {})
    return price_map.get(storage, np.nan)


# Apply release date and launch price
data['Release Date'] = data.apply(map_release_month_pixel, axis=1)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

data['First_Release_Price'] = data.apply(map_launch_price_pixel, axis=1)

# ------------------------------------------------------------
# 5) AGE AT SALE (months)
# ------------------------------------------------------------
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)

# ------------------------------------------------------------
# 6) BUILD REGRESSION DATAFRAME
# ------------------------------------------------------------
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0)
)

df = data.loc[mask].copy()

# price ratio = sold price / launch price
df['Price_ratio'] = df['Price_num'] / df['First_Release_Price']

# storage numeric
df['Storage_num'] = df.apply(lambda r: float(extract_storage(r) or np.nan), axis=1)

df['Year_of_Sale']  = df['Sold Date'].dt.year
df['Month_of_Sale'] = df['Sold Date'].dt.month

# --- Condition cleanup (if column exists) ---
if 'Condition' in df.columns:
    df['Condition'] = (
        df['Condition']
          .astype(str)
          .str.strip()
          .str.title()
    )
    unknowns = {'Unknown', 'Unspecified', '', 'None', 'Na', 'Nan'}
    df.loc[df['Condition'].isin(unknowns), 'Condition'] = np.nan
else:
    df['Condition'] = np.nan  # no column → treat as missing

# Drop rows with NaN in key numeric fields
df = df.dropna(subset=['Age_at_Sale', 'Price_num', 'First_Release_Price', 'Storage_num'])

if df.empty:
    raise ValueError("No valid rows available for Pixel Pro regression after filtering. "
                     "Check Release Date / launch price / storage mapping.")

# --------------------------------------------------------
# 7) ONE-HOT ENCODE CONDITION (IF WE HAVE ANY)
# --------------------------------------------------------
has_any_condition = df['Condition'].notna().any()

if has_any_condition:
    enc = OneHotEncoder(drop=None, sparse_output=False, handle_unknown='ignore')
    cond_mat = enc.fit_transform(df[['Condition']])
    cond_cols = enc.get_feature_names_out(['Condition'])
    cond_df = pd.DataFrame(cond_mat, columns=cond_cols, index=df.index)
else:
    enc = None
    cond_cols = []
    cond_df = pd.DataFrame(index=df.index)

# Numeric features
X_num = df[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']]
X = pd.concat([X_num, cond_df], axis=1)
y = df['Price_ratio']

good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

if X.empty:
    raise ValueError("No rows left after cleaning NaNs in Pixel Pro feature matrix.")

# --------------------------------------------------------
# 8) FIT MODEL
# --------------------------------------------------------
model = lm.LinearRegression()
model.fit(X, y)

print("=== Pixel Pro (Pixel 6 Pro–Pixel 10 Pro XL) Model Fit ===")
print("Samples:", len(y))
print("Intercept:", model.intercept_)
for name, coef in zip(X.columns, model.coef_):
    print(f"{name}: {coef:.6f}")

# --------------------------------------------------------
# 9) PREDICTION HELPER (PIXEL PRO SERIES)
# --------------------------------------------------------
def predict_pixel_pro_price(age_months, storage_gb, year, month, condition, launch_price):
    """
    age_months  : float  (Age_at_Sale in months)
    storage_gb  : int/float (e.g., 128, 256, 512, 1024)
    year        : int   (year of sale, e.g. 2026)
    month       : int   (1–12)
    condition   : str   ('Excellent', 'Very Good', 'Good', 'Fair', etc.)
    launch_price: float (First_Release_Price for that Pixel Pro model + storage)
    """
    cond_str = str(condition).strip().title()

    base = pd.DataFrame({
        'Age_at_Sale':   [age_months],
        'Storage_num':   [float(storage_gb)],
        'Year_of_Sale':  [int(year)],
        'Month_of_Sale': [int(month)],
        'Condition':     [cond_str],
    })

    if enc is not None and len(cond_cols) > 0:
        cond_new = pd.DataFrame(
            enc.transform(base[['Condition']]),
            columns=cond_cols,
            index=base.index
        )
    else:
        cond_new = pd.DataFrame(index=base.index)

    X_new = pd.concat([
        base[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']],
        cond_new
    ], axis=1)

    # align with model columns
    X_new = X_new.reindex(columns=X.columns, fill_value=0)

    ratio = model.predict(X_new)[0]
    return ratio * float(launch_price)

# --------------------------------------------------------
# 10) EXAMPLE USAGE (uncomment to test)
# --------------------------------------------------------
# Example: Predict 12-month-old Pixel 9 Pro 256GB in Excellent condition, sold Jan 2027
# lp = pixel_pro_release_price_map['Pixel 9 Pro']['256']
# pred = predict_pixel_pro_price(12, 256, 2027, 1, "Excellent", lp)
# print(f"Predicted Pixel 9 Pro 256GB resale price: ${pred:.2f}")
```
