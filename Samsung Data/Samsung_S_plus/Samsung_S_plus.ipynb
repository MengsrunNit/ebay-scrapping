```python
import re
import numpy as np
import pandas as pd

# --- Samsung Galaxy S Plus launch prices (US MSRP, S22+–S25+) ---

samsung_plus_release_price_map = {
    # Galaxy S22+ : 128GB $999, 256GB $1049
    'Galaxy S22+': { '128': 999.0, '256': 1049.0 },

    # Galaxy S23+ : base 256GB $999, 512GB $1119
    'Galaxy S23+': { '256': 999.0, '512': 1119.0 },

    # Galaxy S24+ : base 256GB $999, 512GB $1119
    'Galaxy S24+': { '256': 999.0, '512': 1119.0 },

    # Galaxy S25+ : assume same tiers as S24+
    'Galaxy S25+': { '256': 999.0, '512': 1119.0 },
}

# --- Storage extractor (same idea as before) ---

def extract_storage_from_row(row):
    # Try common explicit columns first (if they exist)
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512|1024)', s)
            if m:
                return m.group(1)

    # Fallback: parse Model + Title
    text = f"{row.get('Model', '')} {row.get('Title', '')}"
    m = re.search(r'\b(64|128|256|512|1024)\s*(?:GB|G|gb|Tb|TB)?\b',
                  text, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    return None

# --- Super-simple model key extractor: exact "Galaxy Sxx+" ---

def extract_samsung_plus_model_key(row):
    model = str(row.get('Model', '')).strip()

    # If your Model is exactly "Galaxy S22+" etc:
    for key in samsung_plus_release_price_map.keys():
        if model.startswith(key):
            return key

    # Optional: also look in Title if needed
    title = str(row.get('Title', '')).strip()
    for key in samsung_plus_release_price_map.keys():
        if key in title:
            return key

    return None

# --- Mapper to assign first-release price for Samsung Galaxy S Plus series ---

def map_samsung_plus_first_release_price(row):
    model_key = extract_samsung_plus_model_key(row)   # e.g. "Galaxy S23+"
    if not model_key:
        return np.nan

    storage = extract_storage_from_row(row)
    if not storage:
        return np.nan

    price_map = samsung_plus_release_price_map.get(model_key, {})
    return price_map.get(storage, np.nan)

# --- Load data if needed ---

if 'data' not in globals():
    try:
        data = pd.read_csv('Samsung_plus_base.csv')  # change if needed
        if 'Sold Date' in data.columns:
            data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')
    except FileNotFoundError:
        raise FileNotFoundError('Samsung_plus_base.csv not found; load your Samsung Plus data into `data`')

# Create/overwrite the column with Samsung Plus launch prices
data['First_Release_Price'] = data.apply(map_samsung_plus_first_release_price, axis=1)

# --- Simple sanity check: count non-NaN prices per model string ---

for key in samsung_plus_release_price_map.keys():
    mask = data['Model'].astype(str).str.startswith(key, na=False)
    total = data[mask].shape[0]
    filled = data[mask & data['First_Release_Price'].notna()].shape[0]
    print(f'{key}: filled {filled} / {total}')

# Show sample rows where price was set
display(data[data['First_Release_Price'].notna()].head())
```

```python
# === Samsung S Plus resale regression (Galaxy S22+–S25+) ===

import re
import numpy as np
import pandas as pd
import sklearn.linear_model as lm
from sklearn.preprocessing import OneHotEncoder

# ------------------------------------------------------------
# 1) LOAD / PREP SAMSUNG PLUS DATA
# ------------------------------------------------------------

# If `data` not already defined, load from CSV
if 'data' not in globals():
    data = pd.read_csv('Samsung_S_Plus.csv')  # <-- change filename if needed

# Clean column names and dates
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# Numeric sold price
data['Price_num'] = (
    data['Price']
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
)

# ------------------------------------------------------------
# 2) RELEASE MONTHS FOR S22+–S25+
# ------------------------------------------------------------
release_month_map_plus = {
    '22': "2022-02",  # S22+ Feb 2022
    '23': "2023-02",  # S23+ Feb 2023
    '24': "2024-01",  # S24+ Jan 2024
    '25': "2025-02",  # S25+ early 2025 (assumed)
}

def extract_samsung_plus_model_number(row):
    """
    Extract S number (22, 23, 24, 25) for PLUS models like:
      'Galaxy S22+'
      'Galaxy S23 Plus'
      'Samsung Galaxy S24+ 256GB'
    """
    text = str(row.get('Model', '') or row.get('Title', ''))
    # Fix: handle '+' which is not a word char, so \b fails after it
    m = re.search(
        r'galaxy\s*s\s*(22|23|24|25)\s*(?:\+|plus\b)',
        text,
        flags=re.IGNORECASE
    )
    if m:
        return m.group(1)  # '22', '23', '24', '25'
    return None

def map_release_month_plus(row):
    model_num = extract_samsung_plus_model_number(row)
    if not model_num:
        return np.nan
    return release_month_map_plus.get(model_num.strip(), np.nan)

data['Release Date'] = data.apply(map_release_month_plus, axis=1)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

# ------------------------------------------------------------
# 3) Samsung Plus Launch Prices (US MSRP)
# ------------------------------------------------------------
samsung_plus_launch_price = {
    # S22+ : 128GB $999, 256GB $1049
    '22': { '128': 999.0, '256': 1049.0 },

    # S23+ : base 256GB $999, 512GB $1119 (no 128GB tier)
    '23': { '256': 999.0, '512': 1119.0 },

    # S24+ : base 256GB $999, 512GB $1119
    '24': { '256': 999.0, '512': 1119.0 },

    # S25+ : assume same tiers as S24+
    '25': { '256': 999.0, '512': 1119.0 },
}

def extract_storage(row):
    """Extract storage (GB) from Storage, Model, Title."""
    # direct columns first
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            m = re.search(r'(128|256|512|1024)', str(row[col]))
            if m:
                return m.group(1)

    # fallback: parse from full text
    text = str(row.get('Model', '')) + " " + str(row.get('Title', ''))
    m = re.search(r'\b(128|256|512|1024)\s*(gb|g|tb)?\b', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    return None

def map_launch_price_plus(row):
    model_num = extract_samsung_plus_model_number(row)
    if not model_num:
        return np.nan
    storage = extract_storage(row)
    if not storage:
        return np.nan
    price_map = samsung_plus_launch_price.get(model_num.strip())
    if not price_map:
        return np.nan
    return price_map.get(storage, np.nan)

data['First_Release_Price'] = data.apply(map_launch_price_plus, axis=1)

# ------------------------------------------------------------
# 4) AGE AT SALE (months)
# ------------------------------------------------------------
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)

# ------------------------------------------------------------
# 5) BUILD REGRESSION DATAFRAME
# ------------------------------------------------------------
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0)
)

df = data.loc[mask].copy()

# price ratio = sold price / launch price
df['Price_ratio'] = df['Price_num'] / df['First_Release_Price']

# storage numeric
df['Storage_num'] = df.apply(lambda r: float(extract_storage(r) or np.nan), axis=1)

df['Year_of_Sale']  = df['Sold Date'].dt.year
df['Month_of_Sale'] = df['Sold Date'].dt.month

# --- Condition cleanup (if column exists) ---
if 'Condition' in df.columns:
    df['Condition'] = (
        df['Condition']
          .astype(str).str.strip().str.title()
    )
    unknowns = {'Unknown', 'Unspecified', '', 'None', 'Na', 'Nan'}
    df.loc[df['Condition'].isin(unknowns), 'Condition'] = np.nan
else:
    df['Condition'] = np.nan  # no column → treat as missing

# Drop rows with NaN in key numeric fields
df = df.dropna(subset=['Age_at_Sale', 'Price_num', 'First_Release_Price', 'Storage_num'])

if df.empty:
    raise ValueError("No valid rows available for regression after filtering. Check Release Date / launch price / storage mapping.")

# --------------------------------------------------------
# 6) ONE-HOT ENCODE CONDITION (IF WE HAVE ANY)
# --------------------------------------------------------
has_any_condition = df['Condition'].notna().any()

if has_any_condition:
    enc = OneHotEncoder(drop=None, sparse_output=False, handle_unknown='ignore')
    cond_mat = enc.fit_transform(df[['Condition']])
    cond_cols = enc.get_feature_names_out(['Condition'])
    cond_df = pd.DataFrame(cond_mat, columns=cond_cols, index=df.index)
else:
    enc = None
    cond_cols = []
    cond_df = pd.DataFrame(index=df.index)

# Numeric features
X_num = df[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']]
X = pd.concat([X_num, cond_df], axis=1)
y = df['Price_ratio']

good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

if X.empty:
    raise ValueError("No rows left after cleaning NaNs in feature matrix.")

# --------------------------------------------------------
# 7) FIT MODEL
# --------------------------------------------------------
model = lm.LinearRegression()
model.fit(X, y)

print("=== Samsung S Plus (S22+–S25+) Model Fit ===")
print("Samples:", len(y))
print("Intercept:", model.intercept_)
for name, coef in zip(X.columns, model.coef_):
    print(f"{name}: {coef:.6f}")

# --------------------------------------------------------
# 8) PREDICTION HELPER (PLUS SERIES)
# --------------------------------------------------------
def predict_samsung_plus_price(age_months, storage_gb, year, month, condition, launch_price):
    """
    age_months  : float  (Age_at_Sale in months)
    storage_gb  : int/float (e.g., 256, 512)
    year        : int   (year of sale, e.g. 2026)
    month       : int   (1–12)
    condition   : str   ('Excellent', 'Very Good', 'Good', 'Fair', etc.)
    launch_price: float (First_Release_Price for that S Plus model + storage)
    """
    base = pd.DataFrame({
        'Age_at_Sale':   [age_months],
        'Storage_num':   [storage_gb],
        'Year_of_Sale':  [year],
        'Month_of_Sale': [month],
        'Condition':     [str(condition).strip().title()]
    })

    if enc is not None and len(cond_cols) > 0:
        cond_new = pd.DataFrame(
            enc.transform(base[['Condition']]),
            columns=cond_cols,
            index=base.index
        )
    else:
        cond_new = pd.DataFrame(index=base.index)

    X_new = pd.concat([
        base[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']],
        cond_new
    ], axis=1)

    # align with model columns
    X_new = X_new.reindex(columns=X.columns, fill_value=0)

    ratio = model.predict(X_new)[0]
    return ratio * float(launch_price)

# --------------------------------------------------------
# 9) EXAMPLE USAGE (uncomment to test)
# --------------------------------------------------------
# lp = samsung_plus_launch_price['24']['256']
# pred = predict_samsung_plus_price(12, 256, 2026, 1, "Excellent", lp)
# print(f"Predicted S24+ resale price: ${pred:.2f}")
```

```python
from matplotlib import pyplot as plt
import numpy as np
import os

# [Ensure your 'data' DataFrame is loaded before this point]

# Make sure column names are clean
data.rename(columns=str.strip, inplace=True)

# Use clean columns
x = data['Age_at_Sale']          # months
y = data['Price_num']            # resale price in $

# Drop NaNs/infs before fitting
m = np.isfinite(x) & np.isfinite(y)
x = x[m].to_numpy()
y = y[m].to_numpy()

# -------------------------------
# Convert price -> retention value
# -------------------------------
# Use highest observed price as proxy for launch / max value
original_price = y.max()
retention = (y / original_price) * 100.0   # retention in %

# Fit and build 3rd-degree polynomial on retention
degree = 3
coeffs = np.polyfit(x, retention, degree)
poly_eq = np.poly1d(coeffs)

# Smooth curve for the fit
# Ensure line extends to 48 months for visualization
x_max_plot = max(x.max(), 48)
x_line = np.linspace(x.min(), x_max_plot, 200)
y_line = poly_eq(x_line)

# -------------------------------
# ⭐ PRINT EXACT TABLE VALUES
# -------------------------------
months_to_check = [3, 6, 12, 18, 24, 48]
rmse_val = 0.026222676748056273  # User provided RMSE

print("-" * 40)
print("TABLE VALUES FOR GALAXY S+:")
for month in months_to_check:
    val = poly_eq(month)
    print(f"{month} Months: {val:.2f}%")
print("-" * 40)
print(f"Model RMSE: {rmse_val}")
print("-" * 40)

# Plot scatter + fitted curve (retention)
plt.figure(figsize=(10, 6))
plt.scatter(x, retention, s=15, alpha=0.6, label='Galaxy S+ Model Data')
plt.plot(x_line, y_line, linewidth=2, color='tab:orange', 
         label=f'{degree}rd-degree fit')

# Optional: Add markers for the specific months requested
for month in months_to_check:
    val = poly_eq(month)
    if x.min() <= month <= x_max_plot:
        plt.plot(month, val, 'ro', markersize=5)

# ⭐ Add RMSE Text Box to the plot
textstr = f'RMSE: {rmse_val:.5f}'
props = dict(boxstyle='round', facecolor='white', alpha=0.5)
plt.text(0.02, 0.05, textstr, transform=plt.gca().transAxes, fontsize=10,
        verticalalignment='bottom', bbox=props)

plt.xlabel('Age at Sale (Months)')
plt.ylabel('Value Retained (%)')
plt.title('Galaxy S+ Model Value Retention Over Time')
plt.ylim(0, 110)
plt.legend()
plt.tight_layout()

# Save figure
os.makedirs('output', exist_ok=True)
plt.savefig('output/samsung_s_plus_value_retention_polynomial_fit.png', dpi=300)

plt.show()
```
