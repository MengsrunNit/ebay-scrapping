```python
import re
import pandas as pd

s22ultra = pd.read_csv('galaxy_s22_ultra_only.csv')

# Step 1: Remove rows with '|' (multiple model options)
s22ultra = s22ultra[~s22ultra['Title'].str.contains(r'\|', regex=True, na=False)]

# Step 2: Remove titles that list multiple models like "S22 / S22+" or "S22 - S22 Ultra"
pattern_multi = r'\bS?22\b\s*[-/|]\s*\bS?22\s*(?:Ultra|\+|Plus)?\b'
s22ultra = s22ultra[~s22ultra['Title'].str.contains(pattern_multi, flags=re.IGNORECASE, na=False)]

# Step 3: Keep only "Samsung Galaxy S22 Ultra"
s22ultra = s22ultra[
    s22ultra['Title'].str.contains(r'\b(Samsung\s*)?(Galaxy\s*)?S22\s*Ultra\b', flags=re.IGNORECASE, na=False)
    & ~s22ultra['Title'].str.contains(r'\bS22\s*(\+|Plus|Base|Regular)\b', flags=re.IGNORECASE, na=False)
]

# Preview the results
s22ultra
```

```python
import re
import numpy as np
import pandas as pd

# --- Samsung Galaxy S Ultra launch prices (US MSRP, S22 Ultra–S25 Ultra) ---

samsung_ultra_release_price_map = {
    # Galaxy S22 Ultra
    'Galaxy S22 Ultra': {
        '128': 1199.0,
        '256': 1299.0,
        '512': 1399.0,
        '1024': 1599.0,   # 1TB
    },

    # Galaxy S23 Ultra (no 128GB base)
    'Galaxy S23 Ultra': {
        '256': 1199.0,
        '512': 1379.0,
        '1024': 1619.0,
    },

    # Galaxy S24 Ultra
    'Galaxy S24 Ultra': {
        '256': 1299.0,
        '512': 1419.0,
        '1024': 1659.0,
    },

    # Galaxy S25 Ultra (assumed same as S24 Ultra for now)
    'Galaxy S25 Ultra': {
        '256': 1299.0,
        '512': 1419.0,
        '1024': 1659.0,
    },
}

# --- Storage extractor (same idea as your Plus code) ---

def extract_storage_from_row(row):
    # Try common explicit columns first (if they exist)
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512|1024)', s)
            if m:
                return m.group(1)

    # Fallback: parse Model + Title
    text = f"{row.get('Model', '')} {row.get('Title', '')}"
    m = re.search(
        r'\b(64|128|256|512|1024)\s*(?:GB|G|gb|Tb|TB)?\b',
        text,
        flags=re.IGNORECASE
    )
    if m:
        return m.group(1)

    return None

# --- Super-simple model key extractor: exact "Galaxy Sxx Ultra" ---

def extract_samsung_ultra_model_key(row):
    model = str(row.get('Model', '')).strip()

    # If your Model is exactly "Galaxy S22 Ultra" etc:
    for key in samsung_ultra_release_price_map.keys():
        if model.startswith(key):
            return key

    # Optional: also look in Title if needed
    title = str(row.get('Title', '')).strip()
    for key in samsung_ultra_release_price_map.keys():
        if key in title:
            return key

    return None

# --- Mapper to assign first-release price for Samsung Galaxy S Ultra series ---

def map_samsung_ultra_first_release_price(row):
    model_key = extract_samsung_ultra_model_key(row)   # e.g. "Galaxy S23 Ultra"
    if not model_key:
        return np.nan

    storage = extract_storage_from_row(row)
    if not storage:
        return np.nan

    price_map = samsung_ultra_release_price_map.get(model_key, {})
    return price_map.get(storage, np.nan)

# --- Load data if needed ---

if 'data' not in globals():
    try:
        data = pd.read_csv('Samsung_Series_Ultra_Combined.csv')  # change if needed
        if 'Sold Date' in data.columns:
            data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')
    except FileNotFoundError:
        raise FileNotFoundError('Samsung_Series_Ultra_Combined.csv not found; load your Samsung Ultra data into `data`')

# Create/overwrite the column with Samsung Ultra launch prices
data['First_Release_Price'] = data.apply(map_samsung_ultra_first_release_price, axis=1)

# --- Simple sanity check: count non-NaN prices per Ultra model string ---

for key in samsung_ultra_release_price_map.keys():
    mask = data['Model'].astype(str).str.startswith(key, na=False)
    total = data[mask].shape[0]
    filled = data[mask & data['First_Release_Price'].notna()].shape[0]
    print(f'{key}: filled {filled} / {total}')

# Show sample rows where price was set
display(data[data['First_Release_Price'].notna()].head())
```

```python
# === Samsung S Ultra resale regression (Galaxy S22 Ultra–S25 Ultra) ===

import re
import numpy as np
import pandas as pd
import sklearn.linear_model as lm
from sklearn.preprocessing import OneHotEncoder

# ------------------------------------------------------------
# 1) LOAD / PREP SAMSUNG ULTRA DATA
# ------------------------------------------------------------

# If `data` not already defined, load from CSV
if 'data' not in globals():
    data = pd.read_csv('Samsung_Series_Ultra_Combined.csv')  # <-- change filename if needed

# Clean column names and dates
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# Numeric sold price
data['Price_num'] = (
    data['Price']
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
)

# ------------------------------------------------------------
# 2) RELEASE MONTHS FOR S22 Ultra–S25 Ultra
# ------------------------------------------------------------
release_month_map_ultra = {
    '22': "2022-02",  # S22 Ultra Feb 2022
    '23': "2023-02",  # S23 Ultra Feb 2023
    '24': "2024-01",  # S24 Ultra Jan 2024
    '25': "2025-02",  # S25 Ultra early 2025 (assumed)
}

def extract_samsung_ultra_model_number(row):
    """
    Extract S number (22, 23, 24, 25) for ULTRA models like:
      'Galaxy S22 Ultra'
      'Samsung Galaxy S23 Ultra 256GB'
    """
    text = str(row.get('Model', '') or row.get('Title', ''))
    m = re.search(
        r'galaxy\s*s\s*(22|23|24|25)\s*ultra\b',
        text,
        flags=re.IGNORECASE
    )
    if m:
        return m.group(1)  # '22', '23', '24', '25'
    return None

def map_release_month_ultra(row):
    model_num = extract_samsung_ultra_model_number(row)
    if not model_num:
        return np.nan
    return release_month_map_ultra.get(model_num.strip(), np.nan)

data['Release Date'] = data.apply(map_release_month_ultra, axis=1)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

# ------------------------------------------------------------
# 3) Samsung Ultra Launch Prices (US MSRP)
# ------------------------------------------------------------
# Adjust these if you have an exact launch price table
samsung_ultra_launch_price = {
    # S22 Ultra : 128/256/512/1TB
    '22': { '128': 1199.0, '256': 1299.0, '512': 1399.0, '1024': 1599.0 },

    # S23 Ultra : base 256GB, no 128GB
    '23': { '256': 1199.0, '512': 1379.0, '1024': 1619.0 },

    # S24 Ultra
    '24': { '256': 1299.0, '512': 1419.0, '1024': 1659.0 },

    # S25 Ultra : assume same tiers as S24 Ultra
    '25': { '256': 1299.0, '512': 1419.0, '1024': 1659.0 },
}

def extract_storage(row):
    """Extract storage (GB) from Storage, Model, Title."""
    # direct columns first
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            m = re.search(r'(128|256|512|1024)', str(row[col]))
            if m:
                return m.group(1)

    # fallback: parse from full text
    text = str(row.get('Model', '')) + " " + str(row.get('Title', ''))
    m = re.search(r'\b(128|256|512|1024)\s*(gb|g|tb)?\b', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    return None

def map_launch_price_ultra(row):
    model_num = extract_samsung_ultra_model_number(row)
    if not model_num:
        return np.nan
    storage = extract_storage(row)
    if not storage:
        return np.nan
    price_map = samsung_ultra_launch_price.get(model_num.strip())
    if not price_map:
        return np.nan
    return price_map.get(storage, np.nan)

data['First_Release_Price'] = data.apply(map_launch_price_ultra, axis=1)

# ------------------------------------------------------------
# 4) AGE AT SALE (months)
# ------------------------------------------------------------
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)

# ------------------------------------------------------------
# 5) BUILD REGRESSION DATAFRAME
# ------------------------------------------------------------
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0)
)

df = data.loc[mask].copy()

# price ratio = sold price / launch price
df['Price_ratio'] = df['Price_num'] / df['First_Release_Price']

# storage numeric
df['Storage_num'] = df.apply(lambda r: float(extract_storage(r) or np.nan), axis=1)

df['Year_of_Sale']  = df['Sold Date'].dt.year
df['Month_of_Sale'] = df['Sold Date'].dt.month

# --- Condition cleanup (if column exists) ---
if 'Condition' in df.columns:
    df['Condition'] = (
        df['Condition']
          .astype(str).str.strip().str.title()
    )
    unknowns = {'Unknown', 'Unspecified', '', 'None', 'Na', 'Nan'}
    df.loc[df['Condition'].isin(unknowns), 'Condition'] = np.nan
else:
    df['Condition'] = np.nan  # no column → treat as missing

# Drop rows with NaN in key numeric fields
df = df.dropna(subset=['Age_at_Sale', 'Price_num', 'First_Release_Price', 'Storage_num'])

if df.empty:
    raise ValueError("No valid rows available for Ultra regression after filtering. Check Release Date / launch price / storage mapping.")

# --------------------------------------------------------
# 6) ONE-HOT ENCODE CONDITION (IF WE HAVE ANY)
# --------------------------------------------------------
has_any_condition = df['Condition'].notna().any()

if has_any_condition:
    enc = OneHotEncoder(drop=None, sparse_output=False, handle_unknown='ignore')
    cond_mat = enc.fit_transform(df[['Condition']])
    cond_cols = enc.get_feature_names_out(['Condition'])
    cond_df = pd.DataFrame(cond_mat, columns=cond_cols, index=df.index)
else:
    enc = None
    cond_cols = []
    cond_df = pd.DataFrame(index=df.index)

# Numeric features
X_num = df[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']]
X = pd.concat([X_num, cond_df], axis=1)
y = df['Price_ratio']

good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

if X.empty:
    raise ValueError("No rows left after cleaning NaNs in Ultra feature matrix.")

# --------------------------------------------------------
# 7) FIT MODEL
# --------------------------------------------------------
model = lm.LinearRegression()
model.fit(X, y)

print("=== Samsung S Ultra (S22 Ultra–S25 Ultra) Model Fit ===")
print("Samples:", len(y))
print("Intercept:", model.intercept_)
for name, coef in zip(X.columns, model.coef_):
    print(f"{name}: {coef:.6f}")

# --------------------------------------------------------
# 8) PREDICTION HELPER (ULTRA SERIES)
# --------------------------------------------------------
def predict_samsung_ultra_price(age_months, storage_gb, year, month, condition, launch_price):
    """
    age_months  : float  (Age_at_Sale in months)
    storage_gb  : int/float (e.g., 256, 512, 1024)
    year        : int   (year of sale, e.g. 2026)
    month       : int   (1–12)
    condition   : str   ('Excellent', 'Very Good', 'Good', 'Fair', etc.)
    launch_price: float (First_Release_Price for that Ultra model + storage)
    """
    base = pd.DataFrame({
        'Age_at_Sale':   [age_months],
        'Storage_num':   [storage_gb],
        'Year_of_Sale':  [year],
        'Month_of_Sale': [month],
        'Condition':     [str(condition).strip().str.title()]
    })

    if enc is not None and len(cond_cols) > 0:
        cond_new = pd.DataFrame(
            enc.transform(base[['Condition']]),
            columns=cond_cols,
            index=base.index
        )
    else:
        cond_new = pd.DataFrame(index=base.index)

    X_new = pd.concat([
        base[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']],
        cond_new
    ], axis=1)

    # align with model columns
    X_new = X_new.reindex(columns=X.columns, fill_value=0)

    ratio = model.predict(X_new)[0]
    return ratio * float(launch_price)

# --------------------------------------------------------
# 9) EXAMPLE USAGE (uncomment to test)
# --------------------------------------------------------
# lp = samsung_ultra_launch_price['24']['256']
# pred = predict_samsung_ultra_price(12, 256, 2026, 1, "Excellent", lp)
# print(f"Predicted S24 Ultra resale price: ${pred:.2f}")
```
