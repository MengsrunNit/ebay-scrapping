```python
# Map first release price for iPhone Pro models (12–17) by storage capacity
import re
import numpy as np
import pandas as pd

# Prices (U.S. launch prices) for iPhone Pro models by storage
# Keys are strings like "12 Pro", "13 Pro", ...
release_price_map = {
    # iPhone 12 Pro: 128GB $999, 256GB $1099, 512GB $1299 :contentReference[oaicite:0]{index=0}
    '12 Pro': { '128': 999.0, '256': 1099.0, '512': 1299.0 },

    # iPhone 13 Pro: same launch structure as 12 Pro (Apple kept Pro pricing tiers) :contentReference[oaicite:1]{index=1}
    '13 Pro': { '128': 999.0, '256': 1099.0, '512': 1299.0 },

    # iPhone 14 Pro: same base launch prices as 13 Pro :contentReference[oaicite:2]{index=2}
    '14 Pro': { '128': 999.0, '256': 1099.0, '512': 1299.0 },

    # iPhone 15 Pro: same launch prices (128/256/512) :contentReference[oaicite:3]{index=3}
    '15 Pro': { '128': 999.0, '256': 1099.0, '512': 1299.0 },

    # iPhone 16 Pro: launched at same tiers before being discontinued :contentReference[oaicite:4]{index=4}
    '16 Pro': { '128': 999.0, '256': 1099.0, '512': 1299.0 },

    # iPhone 17 Pro: base storage doubled to 256GB at $1,099 :contentReference[oaicite:5]{index=5}
    '17 Pro': { '256': 1099.0, '512': 1299.0 },
}

# Function to extract storage from common columns or fallback to Title parsing
def extract_storage_from_row(row):
    # Try common explicit columns first (if they exist)
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512)', s)
            if m:
                return m.group(1)

    # Fallback: parse Title (and optionally Model)
    title = str(row.get('Title', ''))
    m = re.search(r'\b(64|128|256|512)\s*(?:GB|G|gb)?\b', title, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    # Try compact forms like '64GB' without space
    m = re.search(r'(64|128|256|512)GB', title, flags=re.IGNORECASE)
    if m:
        return m.group(1)

    return None

# Helper to get model key (e.g., '12 Pro', '15 Pro') from Model or Title fields
def extract_model_key(row):
    # Check Model column first, then Title
    text = str(row.get('Model', '') or row.get('Title', ''))

    # Match patterns like "iPhone 14 Pro", "iphone13 pro", etc.
    m = re.search(r'iphone\s*([0-9]{1,2})\s*pro\b', text, flags=re.IGNORECASE)
    if m:
        num = m.group(1).strip()
        return f'{num} Pro'  # e.g. "12 Pro"
    return None

# Mapper that assigns the release price if we have a mapping for that model+storage
def map_first_release_price(row):
    model_key = extract_model_key(row)   # e.g. "13 Pro"
    if not model_key:
        return np.nan

    if model_key not in release_price_map:
        return np.nan

    storage = extract_storage_from_row(row)
    if storage and storage in release_price_map[model_key]:
        return release_price_map[model_key][storage]

    return np.nan

# Apply mapping to the DataFrame named `data` (already created earlier in the notebook)
# If `data` is not in memory, re-load from the CSV generated earlier: 'iPhone_base.csv'
if 'data' not in globals():
    try:
        data = pd.read_csv('iPhone_base.csv')
        data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')
    except FileNotFoundError:
        raise FileNotFoundError('iPhone_base.csv not found; run earlier cells to produce `data`')

# Create/overwrite the column
data['First_Release_Price'] = data.apply(map_first_release_price, axis=1)

# Quick checks: report filled counts per Pro model
for model_key, mapping in release_price_map.items():
    model_label = f'iPhone {model_key}'  # e.g. "iPhone 13 Pro"
    mask = data['Model'].str.contains(model_label, case=False, na=False)
    total = data[mask].shape[0]
    filled = data[mask & data['First_Release_Price'].notna()].shape[0]
    print(f'{model_label}: filled {filled} / {total}')

# Show sample rows where price was set
display(data[data['First_Release_Price'].notna()].head())

# Optionally save an updated CSV
# data.to_csv('iPhone_Pro_with_release_price.csv', index=False)
```

```python
# === iPhone Pro resale regression (with explicit Condition_Excellent) ===
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder

# ------------------------------------------------------------
# 0) START: assume `data` already exists (DataFrame with at least the columns below)
# Required columns used here:
#   ['Sold Date', 'Release Date' (optional if Age_at_Sale already present),
#    'Age_at_Sale' (months) OR ('Sold Date' & 'Release Date'),
#    'Price_num', 'First_Release_Price', 'Storage' (or Title), 'Condition',
#    'Model' or 'Title' with "Pro" in the name]
# ------------------------------------------------------------

# Basic hygiene
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# -----------------------------
# 0.5) Filter to Pro models
# -----------------------------
if 'Model' in data.columns:
    pro_mask = data['Model'].astype(str).str.contains(r'\bPro\b', case=False, na=False)
elif 'Title' in data.columns:
    pro_mask = data['Title'].astype(str).str.contains(r'\bPro\b', case=False, na=False)
else:
    raise ValueError("Need a 'Model' or 'Title' column to filter Pro devices.")

data = data.loc[pro_mask].copy()

# If Age_at_Sale (months) not present, compute it
if 'Age_at_Sale' not in data.columns:
    data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')
    data['Age_at_Sale'] = (data['Sold Date'] - data['Release Date']).dt.days / 30.44

# Mask of valid rows
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0) &
    data['Sold Date'].notna()
)

df = data.loc[mask].copy()

# Target: normalized resale price (ratio of launch price)
df['Price_ratio'] = (
    pd.to_numeric(df['Price_num'], errors='coerce') /
    pd.to_numeric(df['First_Release_Price'], errors='coerce')
)

# Storage: numeric GB. Prefer 'Storage', fallback to 'Title'
if 'Storage' in df.columns:
    storage_src = df['Storage'].astype(str)
else:
    storage_src = df.get('Title', pd.Series(index=df.index, dtype=str)).astype(str)

df['Storage_num'] = storage_src.str.extract(r'(\d+)', expand=False).astype(float)

# Year / Month of sale
df['Year_of_Sale']  = df['Sold Date'].dt.year
df['Month_of_Sale'] = df['Sold Date'].dt.month

# Condition cleanup: title-case, drop unknown-like
df['Condition'] = (
    df.get('Condition', pd.Series(index=df.index, dtype=object))
      .astype(str)
      .str.strip()
      .str.title()
)
unknown_like = {'Unknown', 'Unspecified', 'Na', 'Nan', 'None', ''}
df.loc[df['Condition'].isin(unknown_like), 'Condition'] = np.nan
df = df.dropna(subset=['Condition'])

# One-hot encode Condition with ALL categories (includes 'Condition_Excellent' explicitly)
enc = OneHotEncoder(drop=None, sparse_output=False, handle_unknown='ignore')
cond_mat  = enc.fit_transform(df[['Condition']])
cond_cols = enc.get_feature_names_out(['Condition'])
cond_df   = pd.DataFrame(cond_mat, columns=cond_cols, index=df.index)

# Assemble feature matrix
X_num = df[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']].apply(
    pd.to_numeric, errors='coerce'
)
X = pd.concat([X_num, cond_df], axis=1)
y = pd.to_numeric(df['Price_ratio'], errors='coerce')

# Final clean for safety
good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

# Fit model (Pro-only)
model = LinearRegression()
model.fit(X, y)

# Report
print("=== Pro Model Fit ===")
print("Samples used:", len(y))
print("Intercept:", model.intercept_)
print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(f"  {name}: {coef:.6f}")

# ------------------------------------------------------------
# Helper: predict absolute resale price in dollars (Pro-only model)
# age_months: float
# storage_gb: int/float (e.g., 128, 256)
# year, month: ints
# condition: string like 'Excellent', 'Very Good', 'Good', 'Fair', 'Poor', etc.
# launch_price: float (First_Release_Price for that Pro model+storage)
# ------------------------------------------------------------
def predict_price_pro(age_months, storage_gb, year, month, condition, launch_price):
    # build a one-row frame
    base = pd.DataFrame({
        'Age_at_Sale': [age_months],
        'Storage_num': [storage_gb],
        'Year_of_Sale': [year],
        'Month_of_Sale': [month],
        'Condition': [str(condition).strip().title()]
    })

    # encode condition using fitted encoder
    cond_new = pd.DataFrame(
        enc.transform(base[['Condition']]),
        columns=cond_cols,
        index=base.index
    )

    X_new = pd.concat(
        [base[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']], cond_new],
        axis=1
    )

    # align columns with training matrix (missing -> 0)
    X_new = X_new.reindex(columns=X.columns, fill_value=0)

    ratio = float(model.predict(X_new))
    return ratio * float(launch_price)

# Example usage (for a Pro device):
# predicted_usd = predict_price_pro(
#     age_months=18,
#     storage_gb=256,
#     year=2024,
#     month=6,
#     condition='Excellent',
#     launch_price=999
# )
# print(f"Predicted Pro resale: ${predicted_usd:.2f}")
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

# ------------------------------------------------------------
# 1) LOAD iPHONE PRO DATA
# ------------------------------------------------------------
data = pd.read_csv('iPhone_Pro.csv')  # change if needed

# Basic cleaning
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# Numeric price
data['Price_num'] = (
    data['Price'].astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
)

# ------------------------------------------------------------
# 2) MAP PRO RELEASE MONTHS (12–17 Pro)
# ------------------------------------------------------------
release_month_map = {
    '12': "2020-11",
    '13': "2021-09",
    '14': "2022-09",
    '15': "2023-09",
    '16': "2024-09",
    '17': "2025-09",
}

def extract_model_number(text):
    m = re.search(r'iphone\s*([0-9]{1,2})', str(text), flags=re.IGNORECASE)
    return m.group(1) if m else None

data['Model_num'] = data.apply(lambda r: extract_model_number(r.get('Model', '') or r.get('Title', '')), axis=1)
data['Release Date'] = data['Model_num'].map(release_month_map)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

# ------------------------------------------------------------
# 3) MAP iPHONE PRO LAUNCH PRICES
# ------------------------------------------------------------
release_price_map = {
    '12': { '128': 999.0,  '256': 1099.0, '512': 1299.0 },
    '13': { '128': 999.0,  '256': 1099.0, '512': 1299.0 },
    '14': { '128': 999.0,  '256': 1099.0, '512': 1299.0 },
    '15': { '128': 999.0,  '256': 1099.0, '512': 1299.0, '1024': 1499.0 },
    '16': { '128': 999.0,  '256': 1099.0, '512': 1299.0, '1024': 1499.0 },
    '17': { '128': 999.0,  '256': 1099.0, '512': 1299.0, '1024': 1499.0 },
}

def extract_storage(row):
    # Prefer Storage column
    if 'Storage' in row and pd.notna(row['Storage']):
        m = re.search(r'(64|128|256|512|1024)', str(row['Storage']))
        if m: return m.group(1)

    # Fallback: parse text
    text = str(row.get('Model', '')) + " " + str(row.get('Title', ''))
    m = re.search(r'\b(64|128|256|512|1024)\b', text)
    return m.group(1) if m else None

def map_price(row):
    mnum = str(row['Model_num'])
    if mnum not in release_price_map:
        return np.nan
    storage = extract_storage(row)
    if storage and storage in release_price_map[mnum]:
        return release_price_map[mnum][storage]
    return np.nan

data['First_Release_Price'] = data.apply(map_price, axis=1)

# ------------------------------------------------------------
# 4) COMPUTE Age_at_Sale
# ------------------------------------------------------------
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)

# ------------------------------------------------------------
# 5) CLEAN + BUILD FINAL DF
# ------------------------------------------------------------
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0)
)

df = data.loc[mask].copy()

df['Price_ratio'] = df['Price_num'] / df['First_Release_Price']

# Extract numeric storage
df['Storage_num'] = df.apply(extract_storage, axis=1).astype(float)

# Clean Condition
df['Condition'] = df['Condition'].astype(str).str.strip().str.title()
df.loc[df['Condition'].isin(['Na', 'Nan', 'None', 'Unspecified', 'Unknown', '']), 'Condition'] = np.nan
df = df.dropna(subset=['Condition'])

# ------------------------------------------------------------
# NUMERIC CONDITION MAPPING (NO ONE-HOT)
# ------------------------------------------------------------
cond_map = {
    'Excellent': 3,
    'Very Good': 2,
    'Good': 1,
    'Fair': 0,
}

df['Condition_num'] = df['Condition'].map(cond_map)
df = df.dropna(subset=['Condition_num'])

# ------------------------------------------------------------
# 6) BUILD FEATURES & FIT MODEL
# ------------------------------------------------------------
X = df[['Age_at_Sale', 'Storage_num', 'Condition_num']].astype(float)
y = df['Price_ratio'].astype(float)

model = LinearRegression()
model.fit(X, y)

print("\n=== iPhone Pro Model Fit (NO One-Hot Condition) ===")
print("Samples:", len(df))
print("Intercept:", model.intercept_)
print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(f"  {name}: {coef:.6f}")

# RMSE
rmse = root_mean_squared_error(y, model.predict(X))
print("RMSE:", rmse)

# ------------------------------------------------------------
# 7) PREDICTION FUNCTION
# ------------------------------------------------------------
def predict_price(age_months, storage_gb, condition, launch_price):
    cond_map = {
        'Excellent': 3,
        'Very Good': 2,
        'Good': 1,
        'Fair': 0,
    }
    cond_val = cond_map.get(str(condition).title())

    base = pd.DataFrame({
        'Age_at_Sale': [age_months],
        'Storage_num': [storage_gb],
        'Condition_num': [cond_val],
    })

    X_new = base[X.columns]  # ensure correct order
    ratio = float(model.predict(X_new))
    return ratio * float(launch_price)

# Example
# predicted = predict_price(18, 256, 'Excellent', 1099)
# print("Predicted resale price:", predicted)
```
