#first release price 

Add a new column `First_Release_Price` for iPhone models based on storage capacity (U.S. launch prices):\n
\n
iPhone 12 (Oct 2020):\n
- 64 GB -> $799\n
- 128 GB -> $849\n
- 256 GB -> $949\n
\n
iPhone 13 (Sept 2021), iPhone 14 (Sept 2022), iPhone 15 (Sept 2023):\n
- 128 GB -> $799\n
- 256 GB -> $899\n
- 512 GB -> $1,099\n
\n
iPhone 16 (Sept 2024):\n
- 128 GB -> $799\n
- 256 GB -> $899\n
- 512 GB -> $1,099\n
\n
iPhone 17 (Sept 2025):\n
- 256 GB -> $799\n
- 512 GB -> $999\n
\n
The code cell below will: try to extract storage from a `Storage`-like column if present,\n
otherwise parse the `Title` column with regex, detect the iPhone model (12/13/14/15/16/17), and then fill `First_Release_Price` accordingly.

```python
# Map first release price for iPhone models (12,13,14,15,16,17) by storage capacity\n
import re
import numpy as np
import pandas as pd

# Prices (U.S. launch prices) by model and storage\n
# Keys are model numbers as strings to make matching simple
release_price_map = {
    '12': { '64': 799.0, '128': 849.0, '256': 949.0 },
    '13': { '128': 799.0, '256': 899.0, '512': 1099.0 },
    '14': { '128': 799.0, '256': 899.0, '512': 1099.0 },
    '15': { '128': 799.0, '256': 899.0, '512': 1099.0 },
    '16': { '128': 799.0, '256': 899.0, '512': 1099.0 },
    '17': { '256': 799.0, '512': 999.0 },
}

# Function to extract storage from common columns or fallback to Title parsing\n
def extract_storage_from_row(row):
    # Try common explicit columns first (if they exist)\n
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512)', s)
            if m:
                return m.group(1)
    # Fallback: parse Title\n
    title = str(row.get('Title', ''))
    m = re.search(r'\b(64|128|256|512)\s*(?:GB|G|gb)?\b', title, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    # Try compact forms like '64GB' without space
    m = re.search(r'(64|128|256|512)GB', title, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

# Helper to get model number (e.g., '12', '13') from Model or Title fields\n
def extract_model_number(row):
    # Check Model column first
    model_field = str(row.get('Model', ''))
    # match 'iPhone' followed by optional spaces and a 1- or 2-digit number (e.g. '12', '16')
    m = re.search(r'iphone\s*([0-9]{1,2})', model_field, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    # Fallback to Title
    title = str(row.get('Title', ''))
    m = re.search(r'iphone\s*([0-9]{1,2})', title, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

# Mapper that assigns the release price if we have a mapping for that model+storage\n
def map_first_release_price(row):
    model_num = extract_model_number(row)
    if not model_num:
        return np.nan
    model_num = model_num.strip()
    if model_num not in release_price_map:
        return np.nan
    storage = extract_storage_from_row(row)
    if storage and storage in release_price_map[model_num]:
        return release_price_map[model_num][storage]
    return np.nan

# Apply mapping to the DataFrame named `data` (already created earlier in the notebook)\n
# If `data` is not in memory, re-load from the CSV generated earlier: 'iPhone_base.csv'\n
if 'data' not in globals():
    try:
        data = pd.read_csv('iPhone_base.csv')
        data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')
    except FileNotFoundError:
        raise FileNotFoundError('iPhone_base.csv not found; run earlier cells to produce `data`')

# Create/overwrite the column
data['First_Release_Price'] = data.apply(map_first_release_price, axis=1)

# Quick checks: report filled counts per model
for model_num, mapping in release_price_map.items():
    model_label = f'iPhone {model_num}'
    total = data[data['Model'].str.contains(model_label, case=False, na=False)].shape[0]
    filled = data[(data['Model'].str.contains(model_label, case=False, na=False)) & (data['First_Release_Price'].notna())].shape[0]
    print(f'{model_label}: filled {filled} / {total}')

# Show sample rows where price was set
display(data[data['First_Release_Price'].notna()].head())

# Optionally save an updated CSV (uncomment to write)\n
# data.to_csv('iPhone_base_with_release_price.csv', index=False)
data
```

```python
# === iPhone resale regression (with explicit Condition_Excellent) ===
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder

# ------------------------------------------------------------
# 0) START: assume `data` already exists (DataFrame with at least the columns below)
# Required columns used here:
#   ['Sold Date', 'Release Date' (optional if Age_at_Sale already present),
#    'Age_at_Sale' (months) OR ('Sold Date' & 'Release Date'),
#    'Price_num', 'First_Release_Price', 'Storage' (or Title), 'Condition']
# ------------------------------------------------------------

# Basic hygiene
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# If Age_at_Sale (months) not present, compute it
if 'Age_at_Sale' not in data.columns:
    data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')
    data['Age_at_Sale'] = (data['Sold Date'] - data['Release Date']).dt.days / 30.44

# Mask of valid rows
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0) &
    data['Sold Date'].notna()
)

df = data.loc[mask].copy()

# Target: normalized resale price (ratio of launch price)
df['Price_ratio'] = pd.to_numeric(df['Price_num'], errors='coerce') / pd.to_numeric(df['First_Release_Price'], errors='coerce')

# Storage: numeric GB. Prefer 'Storage', fallback to 'Title'
if 'Storage' in df.columns:
    storage_src = df['Storage'].astype(str)
else:
    storage_src = df.get('Title', pd.Series(index=df.index, dtype=str)).astype(str)

df['Storage_num'] = storage_src.str.extract(r'(\d+)', expand=False).astype(float)

# Year / Month of sale
df['Year_of_Sale']  = df['Sold Date'].dt.year
df['Month_of_Sale'] = df['Sold Date'].dt.month

# Condition cleanup: title-case, drop unknown-like
df['Condition'] = df.get('Condition', pd.Series(index=df.index, dtype=object)).astype(str).str.strip().str.title()
unknown_like = {'Unknown', 'Unspecified', 'Na', 'Nan', 'None', ''}
df.loc[df['Condition'].isin(unknown_like), 'Condition'] = np.nan
df = df.dropna(subset=['Condition'])

# One-hot encode Condition with ALL categories (includes 'Excellent' explicitly)
enc = OneHotEncoder(drop=None, sparse_output=False, handle_unknown='ignore')
cond_mat  = enc.fit_transform(df[['Condition']])
cond_cols = enc.get_feature_names_out(['Condition'])
cond_df   = pd.DataFrame(cond_mat, columns=cond_cols, index=df.index)

# Assemble feature matrix
X_num = df[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']].apply(pd.to_numeric, errors='coerce')
X = pd.concat([X_num, cond_df], axis=1)
y = pd.to_numeric(df['Price_ratio'], errors='coerce')

# Final clean for safety
good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

# Fit model
model = LinearRegression()
model.fit(X, y)

# Report
print("=== Model Fit ===")
print("Samples used:", len(y))
print("Intercept:", model.intercept_)
print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(f"  {name}: {coef:.6f}")

# ------------------------------------------------------------
# Helper: predict absolute resale price in dollars
# age_months: float
# storage_gb: int/float (e.g., 128, 256)
# year, month: ints
# condition: string like 'Excellent', 'Very Good', 'Good', 'Fair', 'Poor', etc.
# launch_price: float (First_Release_Price for that model+storage)
# ------------------------------------------------------------
def predict_price(age_months, storage_gb, year, month, condition, launch_price):
    # build a one-row frame
    base = pd.DataFrame({
        'Age_at_Sale': [age_months],
        'Storage_num': [storage_gb],
        'Year_of_Sale': [year],
        'Month_of_Sale': [month],
        'Condition': [str(condition).strip().title()]
    })

    # encode condition using fitted encoder
    cond_new = pd.DataFrame(enc.transform(base[['Condition']]), columns=cond_cols, index=base.index)
    X_new = pd.concat([base[['Age_at_Sale','Storage_num','Year_of_Sale','Month_of_Sale']], cond_new], axis=1)

    # align columns with training matrix (missing -> 0)
    X_new = X_new.reindex(columns=X.columns, fill_value=0)

    ratio = float(model.predict(X_new))
    return ratio * float(launch_price)

# Example usage:
# predicted_usd = predict_price(age_months=18, storage_gb=256, year=2024, month=6, condition='Excellent', launch_price=899)
# print(f"Predicted resale: ${predicted_usd:.2f}")
```

```python
# === iPhone resale regression (numeric Condition + RMSE, no year/month) ===
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

# ------------------------------------------------------------
# 0) START: assume `data` already exists (DataFrame with at least the columns below)
# Required columns used here:
#   ['Sold Date', 'Release Date' (optional if Age_at_Sale already present),
#    'Age_at_Sale' (months) OR ('Sold Date' & 'Release Date'),
#    'Price_num', 'First_Release_Price', 'Storage' (or Title), 'Condition']
# ------------------------------------------------------------

# Basic hygiene
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# If Age_at_Sale (months) not present, compute it
if 'Age_at_Sale' not in data.columns:
    data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')
    data['Age_at_Sale'] = (data['Sold Date'] - data['Release Date']).dt.days / 30.44

# Mask of valid rows
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0) &
    data['Sold Date'].notna()
)

df = data.loc[mask].copy()

# Target: normalized resale price (ratio of launch price)
df['Price_ratio'] = (
    pd.to_numeric(df['Price_num'], errors='coerce') /
    pd.to_numeric(df['First_Release_Price'], errors='coerce')
)

# Storage: numeric GB. Prefer 'Storage', fallback to 'Title'
if 'Storage' in df.columns:
    storage_src = df['Storage'].astype(str)
else:
    storage_src = df.get('Title', pd.Series(index=df.index, dtype=str)).astype(str)

df['Storage_num'] = storage_src.str.extract(r'(\d+)', expand=False).astype(float)

# Condition cleanup: title-case, drop unknown-like
df['Condition'] = (
    df.get('Condition', pd.Series(index=df.index, dtype=object))
      .astype(str)
      .str.strip()
      .str.title()
)
unknown_like = {'Unknown', 'Unspecified', 'Na', 'Nan', 'None', ''}
df.loc[df['Condition'].isin(unknown_like), 'Condition'] = np.nan
df = df.dropna(subset=['Condition'])

# ------------------------------------------------------------
# Condition as numeric (NO one-hot encoding)
# ------------------------------------------------------------
cond_map = {
    'Excellent': 3,
    'Very Good': 2,
    'Good': 1,
    'Fair': 0,
}

df['Condition_num'] = df['Condition'].map(cond_map)
df = df.dropna(subset=['Condition_num'])

# Assemble feature matrix: ONLY Age, Storage, Condition_num
X = df[['Age_at_Sale', 'Storage_num', 'Condition_num']].apply(
    pd.to_numeric, errors='coerce'
)
y = pd.to_numeric(df['Price_ratio'], errors='coerce')

# Final clean for safety
good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

# Fit model
model = LinearRegression()
model.fit(X, y)

# Compute RMSE
y_pred = model.predict(X)
rmse = root_mean_squared_error(y, y_pred)

# Report
print("=== Model Fit ===")
print("Samples used:", len(y))
print("RMSE:", rmse)
print("Intercept:", model.intercept_)
print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(f"  {name}: {coef:.6f}")

# ------------------------------------------------------------
# Helper: predict absolute resale price in dollars
# age_months: float
# storage_gb: int/float (e.g., 128, 256)
# condition: string like 'Excellent', 'Very Good', 'Good', 'Fair'
# launch_price: float (First_Release_Price for that model+storage)
# ------------------------------------------------------------
def predict_price(age_months, storage_gb, condition, launch_price):
    cond_map = {
        'Excellent': 3,
        'Very Good': 2,
        'Good': 1,
        'Fair': 0,
    }

    cond_val = cond_map.get(str(condition).strip().title())
    if cond_val is None:
        raise ValueError(f"Unknown condition: {condition}")

    base = pd.DataFrame({
        'Age_at_Sale': [age_months],
        'Storage_num': [storage_gb],
        'Condition_num': [cond_val],
    })

    # Ensure same column order as training
    X_new = base[X.columns]

    ratio = float(model.predict(X_new))
    return ratio * float(launch_price)

# Example usage:
# predicted_usd = predict_price(
#     age_months=18,
#     storage_gb=256,
#     condition='Excellent',
#     launch_price=899
# )
# print(f"Predicted resale: ${predicted_usd:.2f}")
```
