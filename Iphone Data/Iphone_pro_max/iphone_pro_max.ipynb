```python
# Map release month and compute Age_at_Sale for iPhone Pro Max models

import re
import numpy as np
import pandas as pd

# Load data
data = pd.read_csv('iPhone_Pro_Max.csv')

# Ensure Sold Date is datetime
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# Extract year and month of sale
data['year'] = data['Sold Date'].dt.year
data['Month'] = data['Sold Date'].dt.month

# Clean price string -> numeric
data['Price_num'] = (
    data['Price']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

# ===== 1) RELEASE MONTH MAP (12–17 Pro Max) =====
# Use year-month strings like your example
release_month_map = {
    '12': "2020-11",
    '13': "2021-09",
    '14': "2022-09",
    '15': "2023-09",
    '16': "2024-09",
    '17': "2025-09",
}

# ===== 2) Helper: extract model number (12,13,...) from Model/Title =====
def extract_model_number(row):
    # Prefer Model column, fallback to Title
    text = str(row.get('Model', '')) or str(row.get('Title', ''))
    # Match 'iPhone 12 Pro Max', 'iphone13 pro max', etc.
    m = re.search(r'iphone\s*([0-9]{1,2})', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

# ===== 3) Mapper: get release month for that model =====
def map_release_month(row):
    model_num = extract_model_number(row)
    if not model_num:
        return np.nan
    model_num = model_num.strip()
    if model_num in release_month_map:
        return release_month_map[model_num]  # e.g. "2020-11"
    return np.nan

# Create/overwrite the Release Date column as year-month
data['Release Date'] = data.apply(map_release_month, axis=1)

# Convert Release Date to datetime (pandas will assume day=1)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

# ===== 4) Age_at_Sale in months =====
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)
data['Age_at_Sale'] = data['Age_at_Sale'].round(2)

# Optional: quick check
print(data[['Model', 'Sold Date', 'Release Date', 'Age_at_Sale', 'Price', 'Price_num']])
```

```python
# === iPhone Pro Max resale regression (12–17 Pro Max) ===

import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder

# ------------------------------------------------------------
# 1) LOAD PRO MAX DATA
# ------------------------------------------------------------
data = pd.read_csv('iPhone_Pro_Max.csv')

# Basic cleaning
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# Numeric price
data['Price_num'] = (
    data['Price'].astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
)

# ------------------------------------------------------------
# 2) MAP PRO MAX RELEASE MONTHS (12–17 Pro Max)
# ------------------------------------------------------------
# Year-month for Pro Max models
release_month_map = {
    '12': "2020-11",
    '13': "2021-09",
    '14': "2022-09",
    '15': "2023-09",
    '16': "2024-09",
    '17': "2025-09",
}

def extract_model_number(row):
    """Extract iPhone number (12, 13, ...) from Model or Title."""
    text = str(row.get('Model', '') or row.get('Title', ''))
    m = re.search(r'iphone\s*([0-9]{1,2})', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

def map_release_month(row):
    model_num = extract_model_number(row)
    if not model_num:
        return np.nan
    model_num = model_num.strip()
    return release_month_map.get(model_num, np.nan)

data['Release Date'] = data.apply(map_release_month, axis=1)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

# ------------------------------------------------------------
# 3) MAP PRO MAX LAUNCH PRICES (First_Release_Price)
#    (Fill this with your Pro Max launch prices)
# ------------------------------------------------------------
release_price_map = {
    '12': { '128': 1099.0, '256': 1199.0, '512': 1399.0 },
    '13': { '128': 1099.0, '256': 1199.0, '512': 1399.0 },
    '14': { '128': 1099.0, '256': 1199.0, '512': 1399.0 },
    '15': { '256': 1199.0, '512': 1399.0, '1024': 1599.0 },  # example tiers
    '16': { '256': 1199.0, '512': 1399.0, '1024': 1599.0 },
    '17': { '256': 1199.0, '512': 1399.0, '1024': 1599.0 },
}

def extract_storage_from_row(row):
    """Try Storage column first, fallback to Model/Title text."""
    # Prefer an explicit Storage-like column if present
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512|1024)', s)
            if m:
                return m.group(1)

    # Fallback: parse from Model + Title
    text = (str(row.get('Model', '')) + ' ' +
            str(row.get('Title', '')))
    m = re.search(r'\b(64|128|256|512|1024)\s*(?:GB|G|gb|Tb|TB)?\b',
                  text, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

def map_first_release_price(row):
    model_num = extract_model_number(row)
    if not model_num:
        return np.nan
    model_num = model_num.strip()
    if model_num not in release_price_map:
        return np.nan

    storage = extract_storage_from_row(row)
    if storage and storage in release_price_map[model_num]:
        return release_price_map[model_num][storage]
    return np.nan

data['First_Release_Price'] = data.apply(map_first_release_price, axis=1)

# ------------------------------------------------------------
# 4) COMPUTE Age_at_Sale (months)
# ------------------------------------------------------------
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)

# ------------------------------------------------------------
# 5) BUILD REGRESSION DATAFRAME (Pro Max only)
# ------------------------------------------------------------
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0) &
    data['Sold Date'].notna()
)

df = data.loc[mask].copy()

# Target: price / launch price
df['Price_ratio'] = (
    pd.to_numeric(df['Price_num'], errors='coerce') /
    pd.to_numeric(df['First_Release_Price'], errors='coerce')
)

# Storage as numeric GB
if 'Storage' in df.columns:
    storage_src = df['Storage'].astype(str)
else:
    storage_src = df.get('Title', pd.Series(index=df.index, dtype=str)).astype(str)

df['Storage_num'] = storage_src.str.extract(r'(\d+)', expand=False).astype(float)

# Year / month of sale
df['Year_of_Sale']  = df['Sold Date'].dt.year
df['Month_of_Sale'] = df['Sold Date'].dt.month

# Condition cleanup
df['Condition'] = (
    df.get('Condition', pd.Series(index=df.index, dtype=object))
      .astype(str).str.strip().str.title()
)
unknown_like = {'Unknown', 'Unspecified', 'Na', 'Nan', 'None', ''}
df.loc[df['Condition'].isin(unknown_like), 'Condition'] = np.nan
df = df.dropna(subset=['Condition'])

# One-hot encode condition (includes 'Excellent')
enc = OneHotEncoder(drop=None, sparse_output=False, handle_unknown='ignore')
cond_mat  = enc.fit_transform(df[['Condition']])
cond_cols = enc.get_feature_names_out(['Condition'])
cond_df   = pd.DataFrame(cond_mat, columns=cond_cols, index=df.index)

# Feature matrix
X_num = df[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']] \
        .apply(pd.to_numeric, errors='coerce')
X = pd.concat([X_num, cond_df], axis=1)
y = pd.to_numeric(df['Price_ratio'], errors='coerce')

good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

# ------------------------------------------------------------
# 6) FIT LINEAR REGRESSION FOR PRO MAX
# ------------------------------------------------------------
model = LinearRegression()
model.fit(X, y)

print("=== Pro Max Model Fit ===")
print("Samples used:", len(y))
print("Intercept:", model.intercept_)
print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(f"  {name}: {coef:.6f}")


from sklearn.metrics import root_mean_squared_error

# Compute RMSE 
rmse = root_mean_squared_error(___, ____)

# ------------------------------------------------------------
# 7) HELPER: PREDICT PRO MAX RESALE PRICE (ABSOLUTE $)
# ------------------------------------------------------------
def predict_price(age_months, storage_gb, year, month, condition, launch_price):
    """
    age_months : float  (Age_at_Sale in months)
    storage_gb : int/float (e.g., 128, 256, 512)
    year       : int   (year of sale, e.g. 2025)
    month      : int   (1–12)
    condition  : str   ('Excellent', 'Very Good', 'Good', 'Fair', etc.)
    launch_price : float (First_Release_Price for that Pro Max model + storage)
    """
    base = pd.DataFrame({
        'Age_at_Sale': [age_months],
        'Storage_num': [storage_gb],
        'Year_of_Sale': [year],
        'Month_of_Sale': [month],
        'Condition': [str(condition).strip().title()]
    })

    # Encode condition using the fitted encoder
    cond_new = pd.DataFrame(
        enc.transform(base[['Condition']]),
        columns=cond_cols,
        index=base.index
    )
    X_new = pd.concat(
        [base[['Age_at_Sale', 'Storage_num', 'Year_of_Sale', 'Month_of_Sale']], cond_new],
        axis=1
    )

    # Align columns with training matrix (missing -> 0)
    X_new = X_new.reindex(columns=X.columns, fill_value=0)

    ratio = float(model.predict(X_new))
    return ratio * float(launch_price)

# Example usage for a Pro Max:
# iPhone 14 Pro Max, 256 GB, 18 months old, selling in June 2025, Excellent
# launch_price_14pm_256 = release_price_map['14']['256']
# pred_usd = predict_price(age_months=18, storage_gb=256,
#                          year=2025, month=6,
#                          condition='Excellent',
#                          launch_price=launch_price_14pm_256)
# print(f"Predicted resale price: ${pred_usd:.2f}")
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

# ------------------------------------------------------------
# 1) LOAD PRO MAX DATA
# ------------------------------------------------------------
data = pd.read_csv('iPhone_Pro_Max.csv')

# Basic cleaning
data = data.rename(columns=str.strip).copy()
data['Sold Date'] = pd.to_datetime(data['Sold Date'], errors='coerce')

# Numeric price
data['Price_num'] = (
    data['Price'].astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
)

# ------------------------------------------------------------
# 2) MAP PRO MAX RELEASE MONTHS (12–17 Pro Max)
# ------------------------------------------------------------
release_month_map = {
    '12': "2020-11",
    '13': "2021-09",
    '14': "2022-09",
    '15': "2023-09",
    '16': "2024-09",
    '17': "2025-09",
}

def extract_model_number(row):
    """Extract iPhone number (12, 13, ...) from Model or Title."""
    text = str(row.get('Model', '') or row.get('Title', ''))
    m = re.search(r'iphone\s*([0-9]{1,2})', text, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

def map_release_month(row):
    model_num = extract_model_number(row)
    if not model_num:
        return np.nan
    model_num = model_num.strip()
    return release_month_map.get(model_num, np.nan)

data['Release Date'] = data.apply(map_release_month, axis=1)
data['Release Date'] = pd.to_datetime(data['Release Date'], errors='coerce')

# ------------------------------------------------------------
# 3) MAP PRO MAX LAUNCH PRICES (First_Release_Price)
# ------------------------------------------------------------
release_price_map = {
    '12': { '128': 1099.0, '256': 1199.0, '512': 1399.0 },
    '13': { '128': 1099.0, '256': 1199.0, '512': 1399.0 },
    '14': { '128': 1099.0, '256': 1199.0, '512': 1399.0 },
    '15': { '256': 1199.0, '512': 1399.0, '1024': 1599.0 },
    '16': { '256': 1199.0, '512': 1399.0, '1024': 1599.0 },
    '17': { '256': 1199.0, '512': 1399.0, '1024': 1599.0 },
}

def extract_storage_from_row(row):
    """Try Storage column first, fallback to Model/Title text."""
    # Prefer an explicit Storage-like column if present
    for col in ['Storage', 'Capacity', 'Memory', 'Storage Capacity']:
        if col in row.index and pd.notna(row[col]):
            s = str(row[col])
            m = re.search(r'(64|128|256|512|1024)', s)
            if m:
                return m.group(1)

    # Fallback: parse from Model + Title
    text = (str(row.get('Model', '')) + ' ' +
            str(row.get('Title', '')))
    m = re.search(r'\b(64|128|256|512|1024)\s*(?:GB|G|gb|Tb|TB)?\b',
                  text, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None

def map_first_release_price(row):
    model_num = extract_model_number(row)
    if not model_num:
        return np.nan
    model_num = model_num.strip()
    if model_num not in release_price_map:
        return np.nan

    storage = extract_storage_from_row(row)
    if storage and storage in release_price_map[model_num]:
        return release_price_map[model_num][storage]
    return np.nan

data['First_Release_Price'] = data.apply(map_first_release_price, axis=1)

# ------------------------------------------------------------
# 4) COMPUTE Age_at_Sale (months)
# ------------------------------------------------------------
data['Age_at_Sale'] = (
    (data['Sold Date'] - data['Release Date']).dt.days / 30.44
)

# ------------------------------------------------------------
# 5) BUILD REGRESSION DATAFRAME (Pro Max only)
# ------------------------------------------------------------
mask = (
    data['Age_at_Sale'].notna() &
    data['Price_num'].notna() &
    data['First_Release_Price'].notna() &
    (data['First_Release_Price'] > 0) &
    data['Sold Date'].notna()
)

df = data.loc[mask].copy()

# Target: price / launch price
df['Price_ratio'] = (
    pd.to_numeric(df['Price_num'], errors='coerce') /
    pd.to_numeric(df['First_Release_Price'], errors='coerce')
)

# Storage as numeric GB
if 'Storage' in df.columns:
    storage_src = df['Storage'].astype(str)
else:
    storage_src = df.get('Title', pd.Series(index=df.index, dtype=str)).astype(str)

df['Storage_num'] = storage_src.str.extract(r'(\d+)', expand=False).astype(float)

# Condition cleanup
df['Condition'] = (
    df.get('Condition', pd.Series(index=df.index, dtype=object))
      .astype(str).str.strip().str.title()
)
unknown_like = {'Unknown', 'Unspecified', 'Na', 'Nan', 'None', ''}
df.loc[df['Condition'].isin(unknown_like), 'Condition'] = np.nan
df = df.dropna(subset=['Condition'])

# ------------------------------------------------------------
# Encode Condition as an ordered numeric variable
# ------------------------------------------------------------
cond_map = {
    'Excellent': 3,
    'Very Good': 2,
    'Good': 1,
    'Fair': 0,
}

df['Condition_num'] = df['Condition'].map(cond_map)

# Drop rows where we couldn't map the condition
df = df.dropna(subset=['Condition_num'])

# Feature matrix: ONLY age, storage, condition
X = df[['Age_at_Sale', 'Storage_num', 'Condition_num']].apply(
    pd.to_numeric, errors='coerce'
)
y = pd.to_numeric(df['Price_ratio'], errors='coerce')

good = X.notna().all(axis=1) & y.notna()
X = X.loc[good]
y = y.loc[good]

# ------------------------------------------------------------
# 6) FIT LINEAR REGRESSION FOR PRO MAX + RMSE
# ------------------------------------------------------------
model = LinearRegression()
model.fit(X, y)

print("=== Pro Max Model Fit ===")
print("Samples used:", len(y))
print("Intercept:", model.intercept_)
print("Coefficients:")
for name, coef in zip(X.columns, model.coef_):
    print(f"  {name}: {coef:.6f}")

# Compute RMSE 
rmse = root_mean_squared_error(y, model.predict(X))
print("RMSE:", rmse)

# ------------------------------------------------------------
# 7) HELPER: PREDICT PRO MAX RESALE PRICE (ABSOLUTE $)
# ------------------------------------------------------------
def predict_price(age_months, storage_gb, condition, launch_price):
    """
    age_months  : float  (Age_at_Sale in months)
    storage_gb  : int/float (e.g., 128, 256, 512)
    condition   : str or numeric
                  If str: 'Excellent', 'Very Good', 'Good', 'Fair'
    launch_price: float (First_Release_Price for that Pro Max model + storage)
    """
    cond_map = {
        'Excellent': 3,
        'Very Good': 2,
        'Good': 1,
        'Fair': 0,
    }

    if isinstance(condition, str):
        cond_val = cond_map.get(condition.strip().title())
    else:
        cond_val = float(condition)

    base = pd.DataFrame({
        'Age_at_Sale': [age_months],
        'Storage_num': [storage_gb],
        'Condition_num': [cond_val],
    })

    # Ensure same column order as training
    X_new = base.reindex(columns=X.columns, fill_value=0)

    ratio = float(model.predict(X_new))
    return ratio * float(launch_price)

# Example usage for a Pro Max:
# iPhone 14 Pro Max, 256 GB, 18 months old, Excellent
# launch_price_14pm_256 = release_price_map['14']['256']
# pred_usd = predict_price(
#     age_months=18,
#     storage_gb=256,
#     condition='Excellent',
#     launch_price=launch_price_14pm_256
# )
# print(f"Predicted resale price: ${pred_usd:.2f}")
```
